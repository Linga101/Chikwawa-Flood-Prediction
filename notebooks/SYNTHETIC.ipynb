```python
import pandas as pd


data = pd.read_excel("flood_dataset_sample.xlsx")

# simulating 1000 rows
simulated_df = data.sample(n=1000, replace=True)

# resetting index and IDs
simulated_df = simulated_df.reset_index(drop=True)
simulated_df["Pixel_ID"] = range(1, 1001)


simulated_df.to_excel("flood_dataset_simulated_1000.xlsx", index=False)

print(simulated_df.head())
```

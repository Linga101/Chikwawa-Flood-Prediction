# 02. Google Earth Engine Authentication

This notebook covers the one-time authentication process required to run Google Earth Engine (GEE) algorithms and fetch satellite data.

**Pre-requisite:** \n
You must have signed up for a Google Earth Engine account at: https://earthengine.google.com/signup/

```python
import ee

print("Initiating Earth Engine Authentication...")
print("A browser window will pop up. Please log in with your Google account.")
print("Accept the permissions, and copy the authorization token into the prompt if requested.")

# This line opens the browser for OAuth
ee.Authenticate()
```

```python
# After successful authentication above, execute this cell to initialize the Earth Engine API
try:
    ee.Initialize()
    print("Earth Engine initialized successfully! You are ready to fetch satellite data.")
except Exception as e:
    print("Failed to initialize. Make sure you completed the authentication step.")
    print("Error:", e)
```
